```python
# このセルはそのまま実行してください
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 LMの事前学習において、スケーリング則に基づく適切な戦略はどれか?**

1. データサイズを減らしモデルサイズを増やす
2. モデルサイズを最適化しデータサイズを大幅に増加させる
3. データ量が少ない場合でもモデルサイズを無制限に拡大する
4. データとモデルサイズの比率を一定に保ちながら拡大する
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q2　Self-attentionの計算複雑度は系列長$n$を用いてどのように表されるか?**

1. O(n)
2. O($n\sqrt{n}$)
3. O($n^2$)
4. O($n^3$)
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q3 Flash AttentionはどのようにAttentionの高速化を実現したか?**

1. Attentionを高速に近似的に計算する新しい方法を提案した
2. Attentionの計算方法を工夫し、メモリのI/Oの負荷を削減した
3. Attentionを計算する箇所をsparse (疎) にすることで計算量を削減した
4. Attentionの計算を高速化するための新しい方法をTransformerに実装した
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q4　LLMを含む深層学習モデルをレイヤ毎に分割し、分散学習を行う並列方法は次のうちどれか?**

1. パイプライン並列
2. データ並列
3. オプティマイザ並列
4. テンソル並列
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q5　性能を維持しつつ、GPUのメモリ容量に収まるようにLLMのパラメータの量子化を行うことを考えた際、下記のうち正しいものはどれか?**

1. パラメータのビット数を減らすほど、総パラメータ数を増やすことができるため、性能に有利である
2. パラメータは十分な精度がないとあらわせないので、半精度(16ビット)や単精度(32ビット)の浮動小数点数であらわすのが望ましい
3. 性能にはパラメータの浮動小数点数の精度が重要なため、倍精度(64ビット)以上の浮動小数点数であらわすのが望ましい
4. ビット数と総パラメータ数をタスクの性能にあわせて調整することが望ましい
""",
        "options": ["1", "2", "3", "4"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..3 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("LLM_Basic_day5_submission_pred.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'LLM_Basic_day5_submission_pred.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
