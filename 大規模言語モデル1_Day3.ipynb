**以下のセルを修正せずに実行してください。**

```python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd

# --- Q1〜Q4（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 Transformerに関する記述のうち誤った記述はどれ？**

1） あるシーケンス（トークン列）の損失計算を各トークン位置で同時に行うことで学習を並列化できる。
2） RNNやN-gramモデルと比べて、テキストの長距離依存関係をとらえやすい。
3） GPTである一つの文を生成するとき、条件付き確率の推論を各トークン位置で同時に行うことで全位置のトークンを並列に生成できる。
4） 近年のモデルにおいて、Feed Forward層は全体のパラメータの過半数を占めていることが多い。
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q2　TransformerのAttentionに関する記述のうち正しい記述はどれ？**

1）QueryとKeyの次元は異なっていても良い。
2）デコーダのCausal Attentionでは、Queryの位置より先にあるKeyを参照できないようにマスクがされている。
3）Causal Attentionによるマスクが無い一般のAttentionでは、位置$i$に対する位置$j$のAttention scoreを並べた行列$A$は対称行列である。
4）Multi-head AttentionのMulti-headとは、各Transformer BlockでAttentionが繰り返し適用されていることを指している。
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q3 以下の記述の中の[MASK]に入る言葉はどれ？**
位置$i$におけるAttentionの出力を$\mathbf{o}_i$、各位置の[MASK]を $\mathbf{x}_j$、位置$i$に対する位置$j$の（正規化済み）Attention scoreを$a_{ij}$として$\mathbf{o}_i = \sum_j a_{ij} \mathbf{x}_j$である。

1）Query
2）Key
3）Value
4）Output
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q4　次のうち誤った記述はどれ？**

1）近年のLLMは1兆トークンから数十兆トークン程度のテキストデータで事前学習されている。
2）事前学習データを作る方針として、コードや専門書などの混合によるテキストデータの多様化や、WEBデータの収集・加工による高品質化や大規模化などが挙げられる。
3）下流タスクでの性能評価は、訓練データでファインチューニングせずにIn-Context Learningで評価されることがしばしばある。
4）近年のLLMは100epoch程度学習を行う。
""",
        "options": ["1", "2", "3", "4"]
    },
]

# Markdownの行末に半角スペース2つを自動付与
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q4）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q4の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]
        except Exception:
            print("⚠️ 全ての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ 未回答の設問があります。")
            return

        # CSV生成：id=0..3 → Q1〜Q4
        df = pd.Series(answers_mc, name="label")
        df.to_csv("LLM_day3_submission_pred.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'LLM_day3_submission_pred.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
