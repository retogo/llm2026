```python
# このセルはそのまま実行してください（1セル完結）
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 RLHFを行う目的として最も適切なものを1つ選択してください。**

1. 人間のフィードバックをモデルに与えることで、モデルが暴走するのを防ぐため
2. 人間の価値観や意図に沿った出力になるようにモデルを学習するため
3. モデルに新しい知識を追加して、有用性を高めるため
4. モデルの出力のフォーマットを調整するため
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q2 Alignmnetの基準に関する次の説明のうち間違っているものを1つ選択してください。**

1. Helpfulとは、ユーザーの質問に対してできるだけ簡潔で有用な回答を行うという意図を考慮した基準であり、HH-RLHFなどの評価データが存在する
2. Honestとは、情報の虚偽がなく、正確な文章を回答する様な意図を考慮した基準であり、TruthfulQAなどの評価データが存在する
3. Harmlessとは、攻撃的、暴力的なことを回答しない様にするという意図を考慮した基準であり、SHP(Stanford Human Preferences Dataset)などの評価データが存在する
4. Honestとは、事実に基づき虚偽なく回答し、不確実な点はその不確実性を明示するなどしてユーザーを誤解させないことを重視する基準であり、HalEvalなどの評価データが存在する。
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q3 RLVRに関する説明として最も適切なものを１つ選択してください。**
1. 人手でラベル付けしたデータから報酬モデルを学習し、その報酬モデルを用いて強化学習を行う手法。
2. 数学の正解判定やテストケース通過など、自動的に検証可能なシグナルを報酬として用いて強化学習を行う手法。
3. 報酬モデルの学習とベースモデルの教師あり学習を交互に行い、その手順を何度も繰り返す手法。
4. モデル自身が生成した出力を自己採点し、その自己採点スコアを報酬として用いることで方策を更新する手法。
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q4 GRPOのアルゴリズムの説明として間違っているものを1つ選択してください。**

1. 各プロンプトに対して複数の出力を生成し、それらの報酬を比較して相対的に優れた出力に基づくアドバンテージを計算する。
2. 従来のPPOのように価値関数モデルを別途学習せず、グループ内の比較によってアドバンテージを推定する。
3. KL正則化を用いて元のモデルと大きく乖離しすぎないような制約を入れている。
4. GRPOは、グループ内すべてのサンプルに同一のアドバンテージを割り当てることで、サンプル間のばらつきに依存しない安定した学習を行う。
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q5 次の強化学習アルゴリズムのうち、長さバイアス問題の解消を主な目的として設計された手法を1つ選択してください。**

1. GRPO(Group Relative Policy Optimization)
2. DPO(Direct Preference Optimization)
3. DAPO(Decoupled Clip and Dynamic sAmpling Policy Optimization)
4. Dr.GRPO(Dr. Group Relative Policy Optimization)
""",
        "options": ["1", "2", "3", "4"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..4 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("LLM_Basic_day7_submission_pred.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'LLM_Basic_day7_submission_pred.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
