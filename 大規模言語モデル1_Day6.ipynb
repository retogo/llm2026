```python
# このセルはそのまま実行してください（1セル完結）
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 InstructGPT論文で提案された *alignment パイプライン* に含まれない工程は、以下のうちどれ?**

1. Pre-Training
2. In-Context Tuning
3. Supervised Fine-Tuning
4. Reinforcement Learning from Human Feedback
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q2 Instruction Tuningを提案したFLAN論文で報告された、Instruction Tuningの効果として最も適切なものは、以下のうちどれ?**

1. Zero-shot性能の向上
2. Hallucinationの低減
3. 有害な出力の抑制
4. 多言語性能の向上
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q3 LoRAに関する設計指針として、訓練パラメータ総量を一定にした場合により高い性能へつながりやすいのはどれ?**

1. ランク *r* を大きくし、適用層の種類は絞る（例：q のみ）
2. 適用層の種類を増やし（例：q,k,v,o）、ランク *r* は小さくする
3. 出力層のみに適用し、*r*=1 に固定する
4. どの設計でも性能差は出ない
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q4 代表的なPEFT手法の内、入力文の先頭に追加したタスク固有のベクトルを学習するものは、以下のうちどれ?**

1. Adapter
2. Prompt Tuning
3. BitFit
4. LoRA
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q5 各PEFT手法の説明として最も適切なものは、以下のうちどれ?**

1. Adapterでは、Transformer構造を持つAdapterモジュールを追加・学習する
2. Prompt Tuningでは、Soft Promptに加えて、中間層の一部を学習する
3. BitFitでは、Attentionモジュールに含まれるバイアス項のみを学習する
4. LoRAでは、重みパラメータの学習前と学習後の差分を低ランク行列の積とし、それらの低ランク行列を学習する。
""",
        "options": ["1", "2", "3", "4"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..4 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("LLM_Basic_day6_submission_pred.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'LLM_Basic_day6_submission_pred.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
